## Validation Status Summary

The validation registry classifies the implemented experiments according to the level of evidence available from their respective tests.

The categories are:

- **Quantitatively validated** — numerical performance metrics were calculated and used to evaluate the implementation.
- **Validated** — functional behaviour was explicitly tested and confirmed.
- **Validated with limitations** — the main functionality was validated, but documented limitations or unresolved issues remain.
- **Conceptually validated** — the implementation or interface behaviour was demonstrated, but comprehensive quantitative validation was not performed.
- **Proof of concept** — the implementation demonstrated the intended principle on a limited test case.

This classification is intended to distinguish functional demonstration from quantitative performance evaluation.

```python
# 2-D32
# Validation Status Summary

status_counts = {}

for item in validation_registry:
    status = item["status"]
    status_counts[status] = status_counts.get(status, 0) + 1

print("=" * 80)
print("DAY 32 — VALIDATION STATUS SUMMARY")
print("=" * 80)

for status, count in status_counts.items():
    print(f"{status:<42} : {count}")

print()
print(f"Total validation entries: {sum(status_counts.values())}")

expected_total = len(validation_registry)

if sum(status_counts.values()) == expected_total:
    print("Status-count consistency check: PASSED")
else:
    print("Status-count consistency check: FAILED")
```

## Validation Interpretation

The project contains multiple levels of validation evidence.

The strongest quantitative evidence is concentrated in the perception, state-estimation, planning, and vehicle-control experiments where numerical error, accuracy, tracking, or simulation metrics were explicitly calculated.

ROS 2, Gazebo, and JetRacer integration experiments primarily use functional and interface validation. These tests confirm communication, interfaces, execution, and system behaviour, but they do not constitute production-level vehicle validation.

Some experiments remain proof-of-concept implementations. Their results demonstrate the underlying principle but should not be interpreted as evidence of production-level autonomous-driving performance.

## Failure Cases and Known Limitations

The validation review distinguishes between implementation failures and documented project limitations.

The project does not currently contain unresolved implementation errors from the completed daily experiments. However, several experiments have documented limitations that restrict the level of validation that can be claimed.

These limitations include:

- Some perception experiments are proof-of-concept or conceptually validated rather than comprehensively benchmarked.
- Some experiments were validated using limited datasets, scenarios, or test conditions.
- Gazebo and autonomous-vehicle simulation results contain documented simulation limitations.
- The JetRacer vehicle model and integration are conceptually validated rather than physically validated.
- The complete perception-to-vehicle pipeline is not yet implemented as one executable system.
- Standardized message definitions across all major modules are not yet established.
- ROS 2 is not yet the communication backbone for the complete autonomous-driving pipeline.
- MATLAB / Simulink and JetRacer are not connected as one complete hardware-in-the-loop system.
- Physical actuator response and hardware communication latency remain outside the current scope.

These limitations define the boundary of the current validation evidence and should not be interpreted as unresolved implementation failures.

```python
# 3-D32
# Failure Cases and Known Limitations

print("=" * 80)
print("DAY 32 — FAILURE CASES AND KNOWN LIMITATIONS")
print("=" * 80)

follow_up_entries = []

for item in validation_registry:
    status = str(item.get("status", "")).lower()

    if (
        "limitation" in status
        or "proof" in status
        or "conceptual" in status
        or "integration gap" in status
    ):
        follow_up_entries.append(item)

print(f"Validation entries requiring limitation/context review: "
      f"{len(follow_up_entries)}")
print()

for item in follow_up_entries:
    print(
        f"{item.get('day', 'UNKNOWN')} | "
        f"{item.get('name', item.get('title', 'UNKNOWN'))} | "
        f"{item.get('status', 'UNKNOWN')}"
    )

print()
print("Failure-case interpretation:")
print("- No unresolved implementation failure is claimed.")
print("- Entries requiring follow-up are limitation/context cases.")
print("- Conceptual validation is distinguished from quantitative validation.")
print("- System-level integration gaps remain documented.")
print()
print("Failure and limitation review: PASSED")
```

## Reproducibility Review

Reproducibility is evaluated at the level of the documented experiments and validation registry.

The project uses a structured validation record to preserve the relationship between each experiment, its validation status, and the evidence available for that experiment.

For reproducibility, the following aspects are checked:

- Validation entries are present and uniquely identifiable.
- Each validation entry has a documented validation status.
- Validation results are separated from interpretation and limitations.
- Quantitative validation is distinguished from conceptual or proof-of-concept validation.
- Known integration limitations are explicitly documented.
- The validation registry can be inspected independently from the individual experiments.

The reproducibility review does not claim that every experiment can be reproduced identically on different hardware or software environments. It confirms that the current validation evidence and its limitations are documented in a structured and traceable form.

```python
# 4-D32
# Reproducibility Review

print("=" * 80)
print("DAY 32 — REPRODUCIBILITY REVIEW")
print("=" * 80)

total_entries = len(validation_registry)

non_empty_status = all(
    str(item.get("status", "")).strip() != ""
    for item in validation_registry
)

non_empty_identity = all(
    (
        str(item.get("day", "")).strip() != ""
        or str(item.get("id", "")).strip() != ""
    )
    for item in validation_registry
)

days = [
    item.get("day", item.get("id"))
    for item in validation_registry
]

unique_identity = len(days) == len(set(days))

checks = {
    "validation_registry_present": total_entries > 0,
    "validation_entry_count": total_entries == 22,
    "validation_statuses_documented": non_empty_status,
    "validation_entries_identifiable": non_empty_identity,
    "validation_entries_unique": unique_identity,
}

for name, result in checks.items():
    print(f"{name:<40}: {'PASSED' if result else 'FAILED'}")

print("-" * 80)

if all(checks.values()):
    print("Reproducibility review: PASSED")
else:
    print("Reproducibility review: REVIEW REQUIRED")
```
